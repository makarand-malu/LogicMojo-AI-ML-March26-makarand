Task: Train a simple linear regression model using Hours Studied as the independent variable (X) and Marks Scored 
as the dependent variable (y). Report the slope and intercept of the fitted line,
and predict the marks scored by a student who studies for 6 hours.


```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
df = pd.DataFrame([[1,35], [2,45], [3,50], [4,65], [5,70]], columns = ["Hours Studied", "Marks Scored"])
```

```python
df
```

```python
X = df[["Hours Studied"]]
y = df["Marks Scored"]

model = LinearRegression()

model.fit(X, y)
```

```python
# Model Coefficients
slope = model.coef_[0]
intercept = model.intercept_

# 2. Calculate prediction
hours_to_predict = [[6]]
predicted_score = model.predict(hours_to_predict)

# 3. Print the summary
print("=== Linear Regression Model Summary ===")
print(f"Formula: Marks Scored = {slope:.2f} * (Hours Studied) + {intercept:.2f}")
print(f"Slope (Coefficient): {slope:.4f}")
print(f"Intercept: {intercept:.4f}")
print(f"Predictions: {predicted_score[0]:.4f}")
```

**Task**: Fit a polynomial regression model of degree 2 to the given data. 
Report the fitted model equation and predict the value of y when x = 6.

```python
from sklearn.preprocessing import PolynomialFeatures
import numpy as np

# 1. Prepare data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 8, 18, 32, 50])

# 2. Transform X to include x^2 features
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)

# 3. Fit the model
model = LinearRegression()
model.fit(X_poly, y)

# 4. Predict for x = 6
x_new = np.array([[6]])
x_new_poly = poly.transform(x_new)
prediction = model.predict(x_new_poly)

print(f"Fitted Equation: y = {model.coef_[1]:.1f}x² + {model.coef_[0]:.1f}x + {model.intercept_:.1f}")
print(f"Prediction for x=6: {prediction[0]:.1f}")
```

# Problem 6: Z-Score Calculation
A dataset of exam scores is given below.
Scores
45, 50, 55, 60, 65, 70, 75

**Task**: Compute the mean and standard deviation of the dataset. 
Then calculate the z-score for the value 75, and state whether it should be considered an outlier if any value with |z| > 2 is treated as an outlier.

```python
import pandas as pd

# 1. Create the dataset
scores = [45, 50, 55, 60, 65, 70, 75]
df = pd.DataFrame(scores, columns=["Scores"])

# 2. Compute metrics
mean_score = df["Scores"].mean()
std_score = df["Scores"].std()  

print(f"Mean: {mean_score:.2f}")
print(f"Standard Deviation: {std_score:.2f}")

#3. Compute z-score for the value 75
z_score = (75 - mean_score)/std_score

print(f"z-score: {z_score:.2f}")

print("z-score > 2 would cover 95% of the data and typically depends on how strict one wants to be. I prefer 99%")
```

# Problem 8: Binomial Probability
In a quality control process at a manufacturing unit, the probability that a randomly selected item is defective is 0.1. A random sample of 8 items is inspected.
Task: Using the Binomial distribution, calculate the probability that exactly 2 items in the sample of 8 are defective.

```python
from scipy.stats import binom

# Define the binomial parameters
n = 8  # number of items inspected
p = 0.1  # probability of an item being defective
k = 2  # number of defective items we want to find

# Calculate Probability Mass Function (PMF)
probability = binom.pmf(k, n, p)
print(f"Probability of exactly 2 defectives: {probability:.4f}")
```

# Problem 9: Normal Distribution
The scores of students in an examination are normally distributed with a mean of 70 and a standard deviation of 8.

**Task**: (a) What percentage of students scored between 62 and 78? (b) What is the probability that a randomly chosen student scored above 86? Show the calculation using the properties of the normal distribution.

```python
from scipy.stats import norm

mean = 70
std = 8

# Part (a): Cumulative distribution function difference
pct_between = (norm.cdf(78, mean, std) - norm.cdf(62, mean, std)) * 100
print(f"Percentage between 62 and 78: {pct_between:.2f}%")

prob_above = (1-norm.cdf(86, mean, std))*100
print(f"Probability above 86: {prob_above:.4f}%")
```

A company introduced a training program for its employees and recorded their test scores before and after the training.

**Task**: Perform a paired t-test at a 5% significance level to determine whether the training program led to a statistically significant improvement in scores. State the null hypothesis, the alternative hypothesis, compute the t-statistic, and give your conclusion.


```python
from scipy import stats
# 1. Define dataset arrays
before = np.array([70, 72, 68, 75, 74])
after = np.array([78, 80, 76, 82, 79])

# 2. Run a paired, one-sided (greater) t-test
test_results = stats.ttest_rel(after, before, alternative="greater")

print("=== Paired t-Test Results ===")
print(f"t-statistic: {test_results.statistic:.4f}")
print(f"p-value: {test_results.pvalue:.4f}")
```
